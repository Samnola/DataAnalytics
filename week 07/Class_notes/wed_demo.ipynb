## Working with Series in pandas

```python
import pandas as pd
```

```python
grocery_prices = [1.99, 12.99, 3.49, 10.99, 13.50,] 

groc_price_series = pd.Series(grocery_prices)

groc_price_series
```

```python
grocery_prices = {'KitKat':1.99, 'organic eggs':12.99, 'bell pepper': 3.49, 'yogurt (6-pack)':10.99, 'ground beef':13.50,}
```

```python
groc_price_series = pd.Series(grocery_prices)

groc_price_series
```

### Building DataFrame

```python
inventory = {
"product_num": [1658, 2476, 3911],
"quantity": [120, 54, 98]
}

inventory_df = pd.DataFrame(inventory, index = ["KitKat", "bell pepper", "ground beef"])

inventory_df
```
