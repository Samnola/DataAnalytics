## Basic Aggregations

```python
data = {
    'Product': ['A', 'B', 'C', 'A', 'B', 'C', 'A', 'B', 'C'],
    'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North'],
    'Units_Sold': [10, 15, 20, 5, 10, 15, 10, 5, 15],
    'Price': [100, 200, 150, 100, 200, 150, 100, 200, 150]
}
```

```python
df_sales = pd.DataFrame(data)
```

```python
df_sales
```

```python
df_sales["Revenue"] = df_sales["Units_Sold"] * df_sales["Price"]
```

```python
df_sales
```

```python
# Total units sold and total revenue
total_units_sold = df_sales["Units_Sold"].sum()
total_revenue = df_sales["Revenue"].sum()

print("Total Units Sold:", total_units_sold)
print("Total Revenue:", total_revenue)
```

```python
# average units sold and avg revenue
average_units_sold = df_sales["Units_Sold"].mean()
average_revenue = df_sales["Revenue"].mean()

print("Average Units Sold:", average_units_sold)
print("Average Revenue:", average_revenue)
```

## Grouping Data

```python
df_sales.groupby("Product")[["Units_Sold", "Revenue"]].sum()
```

```python
df_sales.groupby("Region")[["Units_Sold", "Revenue"]].mean()
```

```python
df_sales.groupby(["Product", "Region"])[["Units_Sold", "Revenue"]].sum()
```

## Sorting and Ranking Data

```python
df_sales.sort_values(by="Units_Sold")
```

```python
df_sales.sort_values(by="Revenue", ascending=False)
```

```python
df_sales["Revenue_Rank"] = df_sales["Revenue"].rank(ascending=False)

df_sales
```

```python
df_sales["Revenue_Rank"] = df_sales["Revenue"].rank(
    ascending=False,
    method="dense"
)
```

```python
df_sales
```

## Merging DataFrames

```python
product_data = {
    'Product': ['A', 'B', 'C'],
    'Category': ['Electronics', 'Furniture', 'Appliances']
}
```

```python
df_products = pd.DataFrame(product_data)

df_products
```

```python
df_merged = pd.merge(df_sales, df_products, on="Product")

df_merged
```

```python
df_merged.groupby("Category")["Revenue"].sum()
```
