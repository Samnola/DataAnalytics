## Exercise 1.A Numpy: Manipulating Arrays

```python
import numpy as np
```

## Looping Through Arrays

```python
# Create arrA and print
```

```python
arrA = np.array([[11, 12, 13],
                 [14, 15, 16]])

print(arrA)
```

```python
# Loop through the first dimension 
```

```python
for x in arrA:
    for y in x:
        print(y * 2)
    
```

```python
# Difference between loops:
# The first loop multiplies each row by 2.
# The second loop multiplies each individual number by 2.
```

```python
# loop through second dimension again
```

```python
for x in arrA:
    for y in x:
        loop_result = x * 2
        print(loop_result)
```

```python
# This result is different because x is a whole row, 
# x * 2 multiplies the entire row instead of one value.
```

## Joining and Stacking Arrays 

```python
# Concatenate arrA with arrA on axis=0
joined_array_1 = np.concatenate((arrA, arrA), axis=0)

print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1)

print(joined_array_2)
```

```python
# Stack joined_array_1 with itself
stacked_array_1 = np.stack((joined_array_1, joined_array_1))

print(stacked_array_1)
```

```python
# Stack joined_array_1 with joined_array_2
stacked_array_2 = np.stack((joined_array_1, joined_array_2))

print(stacked_array_2)
```

```python
# This causes a ValueError because the arrays do not have the same shape.
# joined_array_1 has shape (4, 3)
# joined_array_2 has shape (2, 6)
# np.stack() only works when all arrays have identical dimensions.
```

```python
# Concatenate joined_array_1 with arrA
another_joined_array = np.concatenate((joined_array_1, arrA))

print(another_joined_array)
```

```python
# Concatenate joined_array_1 with arrA on axis=1
another_2 = np.concatenate((joined_array_1, arrA), axis=1)

print(another_2)

# This gives an error because the arrays do not have the same number of rows.
# joined_array_1 has 4 rows while arrA has 2 rows.
# Arrays must match in size except for the axis being joined.
```

```python
# hstack joins arrays horizontally (side by side)
h_array = np.hstack((arrA, arrA))
print(h_array)

# vstack joins arrays vertically (on top of each other)
v_array = np.vstack((arrA, arrA))
print(v_array)

# dstack stacks arrays depth-wise (creates a third dimension)
d_array = np.dstack((arrA, arrA))
print(d_array)
```

## Splitting Arrays 

```python
# Split joined_array_1 into 2 arrays
splitsky = np.array_split(joined_array_1, 2)

print(splitsky)
```

```python
# Result explanation:
# The result displayed 2 separate arrays.
# Each array is still 2-dimensional.
# Each split array has 2 rows and 3 columns.
```

```python
spliterino = np.array_split(joined_array_1, 3)

print(spliterino)
```

```python
# Result explanation:
# The result contains 3 separate arrays.
# Each array is 2-dimensional.
# Because 4 rows cannot be divided evenly into 3 parts,
# some arrays contain more rows than others.
```
