## Permutations

```python
print(r_4)
```

```python
np.random.shuffle(r_4)

print(r_4)
```

```python
# the rows moved around but the numbers inside each row stayed together.
```

```python
new_shuffle = np.random.permutation(r_4)

print(new_shuffle)

print(r_4)
```

```python
# memorization trick 
# shuffle: mix up array (like shuffling a deck of cards)
# permutation: make me a shuffled copy ( like making a photocopy of the deck first, then shuffling the copy. original stays safe)
```

```python
# Permutation creates a new shuffled version but does not change the original array.
```

shuffle() changes the original array by rearranging its rows. 
permutation() creates a shuffled copy of the array leaving the original array unchanged.

I would use shuffle  when I want to permanently randomize an array, and permutation when I want to keep the original data unchanged.

```python
print(np.add(np.array([1]), r_4))
```

```python
print(np.add(np.array([1,1,1,1,1]), r_4))
```

```python
print(np.add(np.array([[1],[1]]), r_4))
```

```python
# broadcasting error 
```

Broadcasting is when NumPy lets you do math with arrays that don’t have the same shape by stretching the smaller array to match the bigger one.

Using the add function, NumPy adds values together  after it lines up the shapes as best as it can. If one array is smaller or has a single value, it gets repeated across rows or columns so the operation can still work.

The shape of the result depends on the larger array and whether the smaller array can be stretched to match it. If the shapes don’t match and can’t be broadcasted, NumPy will give an error.

## Grocery Problem

```python
# 2 vegetables, 3 fruits, 4 snacks, 0 bev
bag1 = np.array([2, 3, 4, 0])

print(bag1)
```

```python
# 0 fruits, 4 vegetables, 2 bev, 1 snack
bag2 = np.array([4, 0, 1, 2])
print(bag2)
```

```python
total_bag1 = np.sum(bag1)
total_bag2 = np.sum(bag2)

print(f"The first bag contains {total_bag1} items, and the second bag containts {total_bag2} items.")
```

```python
total_categories = bag1 + bag2

print(f"There are {total_categories[0]} vegetables, {total_categories[1]} fruits, {total_categories[2]} boxes of snacks, and {total_categories[3]} beverages in the two grocery bags.")
```

## Teamwork

```python
ali = np.array([5, 4, 6, 4, 2])

print(ali)
```

```python
ben = np.array([0, 5, 5, 3, 4]) # added monday to avoid error message 

print(ben)
```

```python
carla = np.array([3, 6, 0, 3, 4])

print(carla)
```

```python
all_knocks = np.array([ali, ben, carla])

print(all_knocks)
```

```python
daily_total = np.sum(all_knocks, axis=0)
print(daily_total)
```

```python
rolling_knocks = np.cumsum(daily_total)
print(rolling_knocks)
```

```python
estimated_votes = rolling_knocks * 0.5
print(estimated_votes)
```

```python
# Daily total knocks for the whole group
daily_total = all_knocks.sum(axis=0)

# Estimated votes per day 
estimated_votes = daily_total /2

print(estimated_votes)
```
