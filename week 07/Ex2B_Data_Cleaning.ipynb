## Filtering a DataFrame

```python
# rows where age is > 18

df[df["Age"] > 18]
```

```python
# rows where score is > than 50 and < 100

df[(df["Score"] > 50) & (df["Score"] < 100)]
```

## Modifying a DataFrame

```python
df_cleaned = df.drop(df[df["Score"] < 0].index)
```

```python
df_cleaned
```

```python
# Remove city column. Axis 1 tells pandas to drop a column rather than a row

df_cleaned = df_cleaned.drop("City", axis=1)
```

```python
df_cleaned
```

```python
# Replace missing Age value with the mean Age 

df_cleaned["Age"] = df_cleaned["Age"].fillna(df_cleaned["Age"].mean())

df_cleaned
```

```python
# Replace any score value of 0 with the mean score 

mean_score = df_cleaned["Score"].mean()

df_cleaned["Score"] = df_cleaned["Score"].replace(0, mean_score)
```

```python
df_cleaned 
```

```python
# Remove remaining rows containing NaN values 

df_cleaned = df_cleaned.dropna()
```

```python
df_cleaned
```

```python
# Use ID as the index and remove the ID column
df_cleaned = df_cleaned.set_index("ID")
```
