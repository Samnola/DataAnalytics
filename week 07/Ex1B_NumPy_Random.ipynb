```python
import numpy as np
```

## Generating Random Numbers

```python
# random int between 1 and 20, 2 elements

r_1 = np.random.randint(1, 21, size =2) #randint stops before last number 

print(r_1)
```

```python
r_2 = np.random.randint(1, 21, size=(1,2)) 
print(r_2)
```

```python
# random.rand already knows 0 to 1 so only shape is specified 
r_3 = np.random.rand(2,1) 
print(r_3)
```

```python
r_4 = np.random.randint(20, 101, size=(4,5)) #using 101 since 100 needs to be included 

print(r_4)
```

## Generating Random choices from an Array

```python
students = np.array([
    "Naomi", "Corey", "Luis", "Alexus",
    "Victoria", "Samia", "Lesley","Blake",
    "Mohammed", "Charles", "Danait", "Hannah"
])
```

```python
call_on = np.random.choice(students)

print(call_on)
```

```python
# breakout rooms 3 rows, 4 columns
# choice() can repeat names unless you tell it not to

breakout_groups = np.random.choice(students, size=(3,4), replace=False) # by using replace=false it prevents duplicates
```

```python
print(breakout_groups)
```

```python
# I dont believe this would be the best approach to create breakout groups as students can repeat. 
```

```python
# probability values
# Naomi = 30%
# Corey = 0%
# everyone else has remaining 70%
p_values = [
    0.30,  # Naomi
    0.00,  # Corey
    0.07,  # Luis
    0.07,  # Alexus
    0.07,  # Victoria
    0.07,  # Samia
    0.07,  # Lesley
    0.07,  # Blake
    0.07,  # Mohammed
    0.07,  # Charles
    0.07,  # Danait
    0.07   # Hanna
]
```

```python
call_on_weighted = np.random.choice(
    students,
    size=5,
    p=p_values
)

print(call_on_weighted)
```
