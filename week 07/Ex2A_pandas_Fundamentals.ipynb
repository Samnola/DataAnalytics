## Series

```python
aa = [1, 2, 3, 4]

ab = (5, 6, 7, 8)

ac = [9, 10, 11, 12]

ad = {
    'first': 13,
    'second': 14,
    'third': 15,
    
    'fourth': 16
}
```

```python
import pandas as pd
```

```python
ser1 = pd.Series(aa)
ser2 = pd.Series(ab)
ser3 = pd.Series(ac)
ser4 = pd.Series(ad)

print(ser1)
print()

print(ser2)
print()

print(ser3)
print()

print(ser4)
```

```python
# ser5 = pd.Series(17, 18, 19, 20)
# The error happens because I entered the numbers separately instead of putting them together 
# in a list or another collection type before creating the Series.
```

## DataFrames from Series

```python
df1 = pd.DataFrame(ser1) #series is basically a single column of data in pandas. 
                         # data frame is like a spreadsheet/table
```
