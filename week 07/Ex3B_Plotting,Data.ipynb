# Exercise 3.B More plotting practice

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Advanced Plotting Techniques

```python
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23, 17, 35, 29]

plt.bar(products, sales)
plt.show()
```

```python
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23, 17, 35, 29]

plt.bar(products, sales, color='skyblue', width=0.5)

plt.title('Product Sales')
plt.xlabel('Products')
plt.ylabel('Sales')

plt.show()
```

```python
x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x, y)

plt.title('Scatter Plot')
plt.xlabel('X Values')
plt.ylabel('Y Values')

plt.show()
```

```python
x = np.random.rand(50)
y = np.random.rand(50)
s = np.random.rand(50) * 100
colors = np.random.rand(50)

plt.scatter(x, y, s=s, c=colors, cmap='viridis')

plt.title('Customized Scatter Plot')
plt.xlabel('X Values')
plt.ylabel('Y Values')

plt.colorbar(label='Color Scale')

plt.show()
```

## Subplots

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)
```

```python
# Generate 1000 random numbers for the histogram
data = np.random.randn(1000)
```

```python
# Create a figure with two subplots
plt.figure(figsize=(10, 5))
```

```python
# First subplot: Line chart
plt.subplot(1, 2, 1)
plt.plot(x, y)
plt.title('Sine Wave')
plt.xlabel('X Values')
plt.ylabel('sin(x)')
```

```python
# Second subplot: Histogram
plt.subplot(1, 2, 2)
plt.hist(data, bins=30)
plt.title('Histogram of Random Numbers')
plt.xlabel('Value')
plt.ylabel('Frequency')
```

```python
# Adjust layout and display the figure
plt.tight_layout()
plt.show()
```
