# Intro to Numpy

## Creating and working with arrays

```python
import numpy as np
```

```python
x = [1, 2, 3, 4, 5]
y = ('a','b', 'c')
z = {1,2,3, 'a', 'b', 'c'}
zz =[1,2,3, 'a', 'b', 'c']
```

```python
x_arr = np.array(x)
print(x_arr)
print(type(x_arr))
```

```python
print(x)
print(x_arr)
```

```python
y_arr = np.array(y)
print(y_arr)
```

```python
z_arr = np.array(z)
print(z_arr)
```

```python
zz_arr = np.array(zz)
print(zz_arr)
print(zz)
```

```python
t = (('a', 'b'), ('c', 'd')) #tuple containing tuples
```

```python
t_arr = np.array(t) # creates a 2-dimensional array 
print(t_arr)
```

```python
u = (('e','f',),('g', )) # uneven number of values creates an "inhomogenous shape" 
```

```python
u_arr = np.array(u)
print(u_arr)
```

```python
u2 = (('e','f',),('g', None )) # np.nan works better with calculations and can prevent error messages but none will still work.
```

```python
u2_arr = np.array(u2)
print(u2_arr)
```
