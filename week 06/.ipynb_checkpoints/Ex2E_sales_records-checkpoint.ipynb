### Exercise 2.E Using functions for analyzing data
This notebook analyzes a small sales dataset using Python functions.
Each function takes the full dataset as its first argument.

```python
sales_data = [
('Laptop', 'East', 1200.00),
('Phone', 'West', 850.00),
('Tablet', 'East', 400.00),
('Monitor', 'South', 650.00),
('Keyboard', 'West', 75.00),
('Laptop', 'South', 1150.00),
('Phone', 'East', 900.00),
('Tablet', 'West', 425.00),
('Monitor', 'East', 680.00),
('Keyboard', 'South', 80.00),
('Laptop', 'West', 1300.00),
('Phone', 'West', 820.00),
]
```

## Lab 1. Total sales by region

```python
def region_total(data, region):
    return sum(amt for _, r, amt in data if r == region) #the underscore is for variable you are unpacking but do not need to use. 
```

```python
print(region_total(sales_data, 'East'))
print(region_total(sales_data, 'West'))
print(region_total(sales_data, 'South'))
```

## Lab 2. Count of sales by region

```python
def region_count(data, region):
    return sum(1 for _, r, amt in data if r == region) #sum1 counts number of matching records 
```

```python
print(region_total(sales_data, 'East'))
print(region_total(sales_data, 'West'))
print(region_total(sales_data, 'South'))
```

## Lab 3. Average sale amount by region

```python
def region_average(data, region):    
    total = region_total(data, region)  #gets total sales amount for the region
    count = region_count(data, region)   # gets how many records/rows exist for the region

    if count == 0:   #prevents division by 0 if the region does not exist
        return 0

    return total / count 
```

```python
print(region_total(sales_data, 'East'))
print(region_total(sales_data, 'West'))
print(region_total(sales_data, 'South'))
print(region_total(sales_data, 'North'))
```

## Lab 4. Filter by sale amount

```python
def sales_above(data, threshold):
    result = []    #empty list to store approved records 
    for record in data:  #check each rrecord one at a time 
        
        if record[2] > threshold:  #if sale amount is greater than threshold
            result.append(record)
        return result

high_sales = sales_above(sales_data, 500) #testing function using threshold of 500
for record in high_sales:
    print(record)
```

## Lab 5. Top and bottom sale

```python
def top_sale(data):
    return max(data, key=lambda s: s[2])
def bottom_sale(data):
    return min(data, key =lambda s: s[2])

# test the functions
print(top_sale(sales_data))
print(bottom_sale(sales_data))
```

The top sale was a laptop in the West region for $1,300.00

## Lab 6. Apply a discount

```python
def apply_discount(data, discount_pct):
    return list(map(
        lambda s: (s[0], s[1], s[2] * (1-discount_pct)),
        data))
    #test with 15% discount
discounted_sales = apply_discount(sales_data, 0.15)

print("Before Discount:")
for record in sales_data[:3]:
    print(record)
 
    print(record)
```

## lab 7. Unique regions

```python
def unique_regions(data):
    return sorted(set(record[1] for record in data))

#test the function
print(unique_regions(sales_data))
```

This could help organize data and make it easier to identify unique regions without duplicates. 

## Lab 8. Region Summary

```python
def region_summary(data, region):
    # narrowing to only selected region 
    filtered_data = [record for record in data if record[1] == region]

    total = region_total(data, region)
    count = region_count(data, region)
    average = region_average(data,region)
    top = top_sale(filtered_data)

    print(f"=== {region} Region ===")
    print(f"Records: {count}")
    print(f"Total Sales: ${total:,.2f}")
    print(f"Average sale: ${average:,.2f}")
    print(f"Top Sale: {top[0]}-${top[2]:,.2f}")

region_summary(sales_data, "East")
```

## Lab 9. Full summary

```python
for region in unique_regions(sales_data):
    region_summary(sales_data, region)
    print()
```
