## Lab 3. Employee data
This lab exercise makes updates to information extracted from an employee databse.

```python
hr_list = [
    ('0123', 'HR', 'Rebecca Yang', '69000'),
    ('0121', 'IT', 'Mark Blick', '67000'),
    ('0068', 'IT', 'Kahna Larsen', '112000'),
    ('0234','OPS', 'Jim Smith', '54000')
    ]
```

```python
hr_list[1] = ('0121', 'IT', 'Mark Blick-Hawley', '67000')
print(hr_list)
```

```python
hr_list[3] = ('0234', 'CS', 'Jim Smith', '60000')
print(hr_list)
```

```python
for employee in hr_list:
    print(employee[0], "|", employee[1], "|", employee[2], "|", format(int(employee[3]), ","))
```
