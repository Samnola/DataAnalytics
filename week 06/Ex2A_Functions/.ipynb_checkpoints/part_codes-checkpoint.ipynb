# Exercise 2.A Using built-in functions
## Lab 2. Parsing strings - part codes

```python
def get_supplier_code(code):
    return code.split(":")[0]
```

```python
print(get_supplier_code("ACME:123-L"))
```

```python
def get_product_num(num):
    return num.split(":")[1].split("-")[0]
```

```python
print(get_product_num("DI:12345-M"))
```

```python
def get_size(size):
    return size.split("-")[1]
```

```python
print(get_size("ACE:1-12"))
```

```python
part1 = "ACME:123-L"
part2 = "DI:12345-M"
part3 = "ACE:1-12"
```

```python
print("For", part1, ", the_supplier is", get_supplier_code(part1),
      ". Part number", get_product_num(part1),
      "is available in size", get_size(part1))
```
