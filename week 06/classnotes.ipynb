## Working with custom functions
arbitrary arguments

```python
def greeting_kids(*args):
    return args
```

```python
greet = greeting_kids('Henry','Alice')

greet
```

```python
greeting_kids('Victor','Paulo','Henrietta')
```

```python
#keyword arguments

def rels(rel, name, **kwargs):
    kwargs['relationship'] = rel 
    kwargs['name'] = name
    return kwargs
```

```python
person = rels('cousin','Bea', car='Toyotoa Corolla', nickname='gooseface')

print(person)
```

```python
#another example 

def food_card_order(item, **kwargs):
    kwargs['order item'] = item
    return kwargs
```

```python
order1 = food_card_order('hot dog', drink='iced tea', side='fries')

print(order1)
```
