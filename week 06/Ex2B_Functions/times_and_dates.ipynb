Working with datetime functions

```python
import datetime
```

# Today

```python
today = datetime.datetime.now()
```

```python
print(today)
```

```python
print(today.strftime("Today is %A, %B %d, %Y"))
```

```python
# %B Month name, %b short month name, %d Day of month, %Y full year, %y short year caps matter 
```

## My Birthday 

```python
birthday = datetime.date(1996, 2, 27)
print(birthday)
```

# Next 90 Days

```python
# timedelta() = adding/ subtracting time
ninety_d = today + datetime.timedelta(days=90)
print(ninety_d)
```

# Time for Dinner

```python
# time() = creating specific time 
dinner_time = datetime.time(18,30)
print(dinner_time)
```

```python
dinner_time = datetime.time(18, 30)
print("Lets meet for dinner at", dinner_time.strftime("%I:%M %p")) 
# %I hour, %M minutes, %p AM or PM
```
