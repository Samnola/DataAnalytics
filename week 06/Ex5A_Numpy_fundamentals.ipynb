## Array basics

create a regular python list 

```python
fruits = ["apple", "banana", "cherry", "date"]
```

```python
print(fruits)
```

Convert python list into Numpy array 

```python
arr_1 = np.array(fruits)
```

```python
print(arr_1)
```

Check data type of array values

```python
print(arr_1.dtype)
```

The data type returned is unicode text.
The letter "U" stands for unicode string.
The number represents the maximum string length in the array.
