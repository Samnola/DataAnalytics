## Tax functions

This lab demonstrated how to create a function to calculate social securrity tax, medicare tax and federal tax withholdings
based on gross pay and withholding codes. if/elif/else statements were used to determine tax rates based on different withholding codes. I also practiced testing the functions with different employee pay amounts and debugging issues related to function names and indentation. 


```python
def get_soc_sec_tax(gross_pay):
    return gross_pay * 0.062 #assuming tax rate of 6.2%
```

```python
def get_medicare_tax(gross_pay):
    return gross_pay * 0.0145 #assuming tax rate of 1.45%
```

```python
def get_federal_tax(gross_pay, withholding_code):
    if withholding_code == 0:
        tax_rate = 0.23
    elif withholding_code == 1:
        tax_rate =0.21
    elif withholding_code == 2:
        tax_rate =0.195
    elif withholding_code ==3:
        tax_rate = 0.185
    else:
        tax_rate =0.18 - ((withholding_code - 4) * 0.005)
        return gross_pay * tax_rate
```

```python
print(get_soc_sec_tax(1000)) #1000 is simply placeholder waiting for value to be assigned
```

```python
print(get_medicare_tax(1000))
```

```python
print(get_federal_tax(1000, 2))  # the 2 is for the 2 parameters within federal tax
```

```python
# person 1 gross pay $750 and - withholding code 0
print(get_soc_sec_tax(750))
print(get_medicare_tax(750))
print(get_federal_tax(750, 0))
```

```python
# person 2 $1550 gross pay and withholding code 2
print(get_soc_sec_tax(1550))
print(get_medicare_tax(1550))
print(get_federal_tax(1550,2))
```
